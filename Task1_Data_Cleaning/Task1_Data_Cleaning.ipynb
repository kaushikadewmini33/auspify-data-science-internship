```python
print("Shape before:", df.shape)
df = df.drop_duplicates(subset=['title', 'type', 'release_year', 'director'], keep='first')
print("Shape after:", df.shape)
print("Duplicate title+type:", df.duplicated(subset=['title', 'type']).sum())
```

```python
#Step 3: Handle null values and inconsistent formats

import numpy as np
for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()
```

```python
df = df.replace('Not Given', np.nan)
df.isnull().sum()
```

```python
df['director'] = df['director'].fillna('Unknown')
df['country'] = df['country'].fillna('Unknown')
df.isnull().sum()
```

```python
df['rating'] = df['rating'].replace({'NR': 'Not Rated', 'UR': 'Not Rated'})
df['rating'].value_counts()
```

```python
print(df.isnull().sum().sum(), "nulls left")
print("Director Unknown:", (df['director'] == 'Unknown').sum())
print("Country Unknown:", (df['country'] == 'Unknown').sum())
print("'Not Given' left:", (df == 'Not Given').sum().sum())
```

```python
# Step 4: Transform categorical and date-related columns

df['date_added'] = pd.to_datetime(df['date_added'], format='%m/%d/%Y')
df['year_added'] = df['date_added'].dt.year
df['month_added'] = df['date_added'].dt.month
df['month_name'] = df['date_added'].dt.month_name()
df['day_of_week'] = df['date_added'].dt.day_name()
df[['date_added', 'year_added', 'month_added', 'month_name', 'day_of_week']].head()
```

```python
df['duration_value'] = df['duration'].str.extract(r'(\d+)').astype(int)
df['duration_unit'] = df['duration'].str.extract(r'([A-Za-z]+)')
df['duration_unit'] = df['duration_unit'].replace({'Seasons': 'Season'})
df[['duration', 'duration_value', 'duration_unit']].head()
df['duration_unit'].value_counts()
```

```python
df[['duration', 'duration_value', 'duration_unit']].head()
```

```python
df['primary_country'] = df['country'].str.split(',').str[0].str.strip()
df['primary_genre'] = df['listed_in'].str.split(',').str[0].str.strip()
df[['country', 'primary_country', 'listed_in', 'primary_genre']].head()
```

```python
rating_map = {
    'TV-Y': 'Kids', 'TV-Y7': 'Kids', 'TV-Y7-FV': 'Kids', 'TV-G': 'Kids', 'G': 'Kids',
    'TV-PG': 'Older Kids', 'PG': 'Older Kids',
    'TV-14': 'Teens', 'PG-13': 'Teens',
    'TV-MA': 'Adults', 'R': 'Adults', 'NC-17': 'Adults',
    'Not Rated': 'Not Rated'
}
df['audience'] = df['rating'].map(rating_map)
print("Unmapped ratings:", df['audience'].isnull().sum())
df['audience'].value_counts()
```

```python
for col in ['type', 'rating', 'duration_unit', 'audience']:
    df[col] = df[col].astype('category')
df.dtypes
```

```python
# Step 5: Create the clean dataset

df.info()
print(df.isnull().sum().sum(), "missing values left")
print("Duplicates:", df.duplicated().sum())
print("Duplicate show_id:", df['show_id'].duplicated().sum())
print("Final shape:", df.shape)
```
